```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

# Ask the user to enter a temperature
celsius = float(input("Enter temperature in Celsius: "))

# Convert Celsius to Fahrenheit
fahrenheit = celsius_to_fahrenheit(celsius)

# Display the result
print("Temperature in Fahrenheit:", fahrenheit)

# Set the temperature threshold
threshold = 30

# Check whether the temperature is above or below the threshold
if celsius > threshold:
    print("The temperature is above 30°C.")
elif celsius < threshold:
    print("The temperature is below 30°C.")
else:
    print("The temperature is exactly 30°C.")
# Convert several temperatures using a loop
print("\nConverting several temperatures:")

temperatures = [0, 10, 20, 30, 40, 60]

for temp in temperatures:
    converted = celsius_to_fahrenheit(temp)
    print(f"{temp}°C = {converted:.2f}°F")
```

```python
This is a code that enables the conversion of temperature from degree Celsius to Farenheit with a threshold of 30°c. It gives a response if the temperature is higher, equal or lower than 30°c
```
